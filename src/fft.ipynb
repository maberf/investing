```python
import numpy as np
from scipy import fftpack
import matplotlib.pyplot as plt

time_step = 0.05
time_vec = np.arange (0, 10, time_step)
period = 5
sig = (np.sin(2*np.pi*time_vec/period) + 0.25*np.random.randn(time_vec.size))

print(np.round(sig,2))
```

```python
plt.plot(np.round(sig,2))
```

```python
sig_fft = fftpack.fft(sig)
```

```python
amplitude = np.abs(sig_fft)
```

```python
angle = np.angle(sig_fft)
```

```python
frequency = fftpack.fftfreq(sig.size, d=time_step)
```

```python
print(len(amplitude))
print(amplitude)
```

```python
print(len(frequency))
print(frequency)
```

```python
# Criação do gráfico
plt.figure(figsize=(10, 6))
plt.plot(frequency, amplitude, marker='o', linestyle='-', color='b', label='Signal')

# Formatação e Identificação
plt.title('FFT Frequency Distribution')
plt.xlabel('Frequency')
plt.ylabel('Amplitude')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Exibição
plt.show()
```

```python
ampl_freq = np.array([amplitude, frequency])
ampl_position = ampl_freq[0, :].argmax()
peak_freq = ampl_freq[1, ampl_position]
```

```python
print(ampl_freq.shape[0])
print(ampl_freq.shape[1])
print(ampl_freq.size)
print(ampl_freq)
```

```python
print(ampl_position)
print(peak_freq)
```

```python
high_freq_fft = sig_fft.copy()
high_freq_fft[np.abs(frequency) > peak_freq] = 0
```

```python
print(high_freq_fft)
```

```python
filtered_sig = fftpack.ifft(high_freq_fft)
print(filtered_sig)
```

```python
plt.plot(np.round(filtered_sig,2))
```
